```python
def addDays(data: str, dia: int) -> str:
    """
    Retorna uma nova data, a partir da data inicial informada, acrescida de uma quantidade de dias,
    considerando a transição entre meses e anos quando o total ultrapassa o último dia do mês.

    Parâmetros:
        data (str): Data inicial no formato 'YYYY-MM-DD'.
        dia (int): Número de dias a serem adicionados.

    Retorna:
        str: Data resultante no formato 'YYYY-MM-DD'.
    """
    date_obj = datetime.strptime(data, "%Y-%m-%d")
    new_date = date_obj + timedelta(days=dia)
    return new_date.strftime("%Y-%m-%d")
```
