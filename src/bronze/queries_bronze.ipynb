# Projeto: Criação da Estrutura das Tabelas Bronze

### Objetivo  
Executar a **criação da estrutura (esqueleto)** das tabelas da camada **Bronze**, destinadas ao armazenamento padronizado de previsões climáticas por fonte de origem.

---

### Escopo  
Este notebook **não realiza inserção de dados**. Ele apenas monta e estrutura através do **`CREATE TABLE IF NOT EXISTS`** para garantir que as estruturas existam previamente no catálogo do Databricks.

---

### Tabelas Criadas

| Tabela                          | Fonte de Dados     |
|--------------------------------|--------------------|
| `grupo01.bronze.clima_tempo`   | Site ClimaTempo         |
| `grupo01.bronze.inpe`          | API INPE               |
| `grupo01.bronze.open_weather_map` | API OpenWeatherMap  |
| `grupo01.bronze.weather`       | Site Weather.com |
| `grupo01.bronze.weather_api`       | API Weather_api |


---

### Observação  
As tabelas são criadas no formato **Delta Lake**, com schemas previamente definidos e **sem nenhuma carga inicial de dados**.  
Elas servem como base para futuras etapas de ingestão e transformação de dados estruturados.

```python
create_table_clima_tempo = """
    CREATE TABLE IF NOT EXISTS grupo01.bronze.clima_tempo (
        data_carga STRING COMMENT 'Data em que os dados foram carregados',
        data_previsao STRING COMMENT 'Data da previsão do clima',
        cidade STRING COMMENT 'Nome da cidade',
        estado STRING COMMENT 'Nome do estado',
        temperatura_maxima STRING COMMENT 'Temperatura máxima prevista',
        temperatura_minima STRING COMMENT 'Temperatura mínima prevista',
        umidade_maxima STRING COMMENT 'Umidade máxima prevista',
        umidade_minima STRING COMMENT 'Umidade mínima prevista',
        precipitacao STRING COMMENT 'Quantidade de precipitação prevista e chance de ocorrer',
        vento STRING COMMENT 'Velocidade do vento prevista e direção',
        fonte STRING COMMENT 'Fonte dos dados'
    ) 

    USING DELTA
    COMMENT 'Tabela contendo previsões climáticas das capitais brasileiras da fonte clima_tempo';
"""
```

```python
create_table_open_weather_map = """
    CREATE TABLE IF NOT EXISTS grupo01.bronze.open_weather_map (
        data_carga STRING COMMENT 'Data em que os dados foram carregados',
        data_previsao STRING COMMENT 'Data da previsão do clima',
        cidade STRING COMMENT 'Nome da cidade',
        estado STRING COMMENT 'Sigla do estado',
        temperatura_maxima_dia STRING COMMENT 'Temperatura máxima prevista para o dia',
        temperatura_minima_dia STRING COMMENT 'Temperatura mínima prevista para o dia',
        umidade_maxima_dia STRING COMMENT 'Umidade máxima prevista para o dia',
        umidade_minima_dia STRING COMMENT 'Umidade mínima prevista para o dia',
        precipitacao STRING COMMENT 'Precipitação prevista em mm',
        vel_vento_media_km STRING COMMENT 'Velocidade média do vento (km/h)',
        fonte STRING COMMENT 'Fonte dos dados'
    )
    USING DELTA
    COMMENT 'Tabela contendo previsões climáticas diárias das capitais brasileiras, extraídas da API OpenWeatherMap';
"""
```
