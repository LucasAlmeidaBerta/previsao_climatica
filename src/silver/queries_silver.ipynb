# Projeto: Criação da Estrutura das Tabelas Silver

### Objetivo  
Executar a **criação da estrutura (esqueleto)** das tabelas da camada **Silver**, destinadas ao armazenamento padronizado de previsões climáticas por fonte de origem.

---

### Escopo  
Este notebook **não realiza inserção de dados**. Ele apenas monta e estrutura através do **`CREATE TABLE IF NOT EXISTS`** para garantir que as estruturas existam previamente no catálogo do Databricks.

---

### Tabelas Criadas

| Tabela                          | Fonte de Dados     |
|--------------------------------|--------------------|
| `grupo01.silver.clima_tempo`   | ClimaTempo         |
| `grupo01.silver.inpe`          | INPE               |
| `grupo01.silver.open_weather_Map` | OpenWeatherMap  |
| `grupo01.silver.weather`       | Weather |
| `grupo01.silver.weather_api`       | Weather API |


---

### Observação  
As tabelas são criadas no formato **Delta Lake**, com schemas previamente definidos e **sem nenhuma carga inicial de dados**.  
Elas servem como base para futuras etapas de ingestão e transformação de dados estruturados.

```python
create_table_silver_clima_tempo = """
    CREATE TABLE IF NOT EXISTS grupo01.silver.clima_tempo (
    id STRING COMMENT 'Chave primária gerada por hash SHA-256 com base em dt_previsao, nom_municipio, sig_uf e flg_fonte_previsao',
    dt_carga TIMESTAMP COMMENT 'Data em que os dados foram carregados formato dd/mm/yyyy  HH:mm:ss',
    dt_previsao DATE COMMENT 'Data da previsão do clima formato dd/mm/yyyy',
    nom_municipio STRING COMMENT 'Nome da cidade',
    sig_uf STRING COMMENT 'Sigla do estado (UF)',
    temp_max DECIMAL(5,2) COMMENT 'Temperatura máxima prevista em graus celsius (C°)',
    temp_min DECIMAL(5,2) COMMENT 'Temperatura mínima prevista em graus celsius (C°)',
    umid_max DECIMAL(5,2) COMMENT 'Umidade máxima prevista em porcentagem (%)',
    umid_min DECIMAL(5,2) COMMENT 'Umidade mínima prevista em porcentagem (%)',
    precipitacao DECIMAL(5,2) COMMENT 'Quantidade de precipitação prevista em milímetros (mm/h)',
    prob_precipitacao DECIMAL(5,2) COMMENT 'Chance de ocorrer precipitação em porcentagem (%)',
    vel_vento DECIMAL(5,2) COMMENT 'Velocidade do vento prevista em quilomêtros por hora (km/h)',
    flg_fonte_previsao STRING COMMENT 'Fonte dos dados'
) USING DELTA
COMMENT 'Tabela contendo previsões climáticas das capitais brasileiras da fonte clima_tempo';
"""
```

```python
create_table_silver_inpe = """ 
    CREATE TABLE IF NOT EXISTS grupo01.silver.inpe (
        id STRING COMMENT 'Chave primária gerada por hash SHA-256 com base em dt_previsao, nom_municipio, sig_uf e flg_fonte_previsao',
        dt_carga TIMESTAMP COMMENT 'Data em que os dados foram carregados formato dd/mm/yyyy HH:mm:ss',
        dt_previsao DATE COMMENT 'Data da previsão do clima formato dd/mm/yyyy',
        nom_municipio STRING COMMENT 'Nome da cidade',
        sig_uf STRING COMMENT 'Sigla do estado (UF)',
        temp_max DECIMAL(5,2) COMMENT 'Temperatura máxima prevista em graus celsius (C°)',
        temp_min DECIMAL(5,2) COMMENT 'Temperatura mínima prevista em graus celsius (C°)',
        precipitacao DECIMAL(5,2) COMMENT 'Quantidade de precipitação prevista em milímetros (mm/h)',
        prob_precipitacao DECIMAL(5,2) COMMENT 'Chance de ocorrer precipitação em porcentagem (%)',
        flg_fonte_previsao STRING COMMENT 'Fonte dos dados'    
    ) USING DELTA
    COMMENT 'Tabela contendo previsões climáticas das capitais brasileiras da fonte inpe';
"""
```

```python
create_table_open_weather_map = """
    CREATE TABLE IF NOT EXISTS grupo01.silver.open_weather_map (
        id STRING COMMENT 'Chave primária gerada por hash SHA-256 com base em dt_previsao, nom_municipio, sig_uf e flg_fonte_previsao',
        dt_carga TIMESTAMP COMMENT 'Data em que os dados foram carregados no formato dd/mm/yyyy HH:mm:ss',
        dt_previsao DATE COMMENT 'Data da previsão do clima no formato dd/mm/yyyy',
        nom_municipio STRING COMMENT 'Nome do município (cidade)',
        sig_uf STRING COMMENT 'Sigla da unidade federativa (UF)',
        temp_max DECIMAL(5,2) COMMENT 'Temperatura máxima prevista em graus Celsius (°C)',
        temp_min DECIMAL(5,2) COMMENT 'Temperatura mínima prevista em graus Celsius (°C)',
        umid_max DECIMAL(5,2) COMMENT 'Umidade relativa do ar máxima prevista (%)',
        umid_min DECIMAL(5,2) COMMENT 'Umidade relativa do ar mínima prevista (%)',
        precipitacao DECIMAL(5,2) COMMENT 'Quantidade total de precipitação prevista para o dia (mm)',
        vel_vento DECIMAL(5,2) COMMENT 'Velocidade média do vento prevista (km/h)',
        flg_fonte_previsao STRING COMMENT 'Fonte dos dados meteorológicos'
    )
    USING DELTA
    COMMENT 'Tabela contendo previsões climáticas das capitais brasileiras provenientes da fonte OpenWeatherMap';
"""
```
