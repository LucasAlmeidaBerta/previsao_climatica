# Projeto: Coleta de Previsão Climática – Weather API

### Descrição Geral
Este documento descreve o processo de coleta de dados meteorológicos via [Weather API]("http://api.weatherapi.com/v1/forecast.json"), utilizando o ambiente Databricks.<br>
O objetivo é extrair previsões climáticas para todas as capitais brasileiras, no período de 15 dias, estruturando as informações para posterior análise e tratamento em pipelines de dados.

---

### Informações Técnicas

- **Fonte de Dados**: [Weather API – Previsão de 15 dias]("http://api.weatherapi.com/v1/forecast.json")
- **Tecnologias Utilizadas**: 
  - Python (Requests, BeautifulSoup)
  - Apache Spark (PySpark)
  - Databricks Notebooks

- **Dados Coletados**:
  - Cidade e Unidade Federativa (UF)
  - Data da previsão
  - Temperatura mínima e máxima
  - Umidade mínima e máxima
  - Precipitação e Probabilidade de chuva
  - Velocidade do vento
  - Umidade mínima e máxima
