# Projeto: Coleta de Previsão Climática – Open Weather Map API

### Descrição Geral
Este documento descreve o processo de coleta de dados meteorológicos via [Open Weather Map]("https://openweathermap.org/api/one-call-3"), utilizando o ambiente Databricks.<br>
Faz a coleta e salva os registros em um arquivo parquet com previsões do tempo para os próximos 10 dias nas 27 capitais do Brasil, usando a API One Call 3.0 (endpoint day_summary) do OpenWeather.

---

### Informações Técnicas

- **Fonte de Dados**: [One Call API 3.0 – Previsão de 10 dias]("https://openweathermap.org/api/one-call-3")
- **Tecnologias Utilizadas**: 
  - Python (Requests, BeautifulSoup)
  - Apache Spark (PySpark)
  - Databricks Notebooks

- **Dados Coletados**:
  - Cidade e Unidade Federativa (UF)
  - Data da previsão
  - Temperatura mínima e máxima
  - Umidade mínima e máxima
  - Precipitação e Probabilidade de chuva
  - Velocidade do vento
  - Umidade mínima e máxima
