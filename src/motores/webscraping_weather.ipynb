# Projeto: Coleta de Previsão Climática – Weather

### Descrição Geral
Este documento descreve o processo de coleta de dados meteorológicos via *web scraping* no site [Weather.com](https://weather.com/pt-BR/clima/hoje/l/BRXX0043:1:BR?Goto=Redirected), utilizando o ambiente Databricks.<br>
O objetivo é extrair previsões climáticas para todas as capitais brasileiras, no período de 7 dias, estruturando as informações para posterior análise e tratamento em pipelines de dados.

---

### Informações Técnicas

- **Fonte de Dados**: [Weather – Previsão de 7 dias](https://weather.com/pt-BR/clima/hoje/l/BRXX0043:1:BR?Goto=Redirected)
- **Tecnologias Utilizadas**: 
  - Python (Requests, BeautifulSoup)
  - Apache Spark (PySpark)
  - Databricks Notebooks

- **Dados Coletados**:
  - Cidade e Unidade Federativa (UF)
  - Data da previsão
  - Temperatura mínima e máxima
  - Probabilidade de chuva
  - Velocidade do vento
  - Umidade mínima e máxima
