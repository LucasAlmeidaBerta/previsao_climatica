# Projeto: Coleta de Previsão Climática – ClimaTempo

### Descrição Geral
Este documento descreve o processo de coleta de dados meteorológicos via *web scraping* no site [ClimaTempo](https://www.climatempo.com.br/previsao-do-tempo), utilizando o ambiente Databricks.

O objetivo é extrair previsões climáticas para todas as capitais brasileiras, no período de 15 dias, estruturando as informações para posterior análise e tratamento em pipelines de dados.

---

### Informações Técnicas

- **Fonte de Dados**: [ClimaTempo – Previsão de 15 dias](https://www.climatempo.com.br/previsao-do-tempo)
- **Tecnologias Utilizadas**: 
  - Python (Requests, BeautifulSoup)
  - Apache Spark (PySpark)
  - Databricks Notebooks

- **Dados Coletados**:
  - Cidade e Unidade Federativa (UF)
  - Data da previsão
  - Temperatura mínima e máxima
  - Probabilidade de chuva
  - Velocidade do vento
  - Umidade mínima e máxima
