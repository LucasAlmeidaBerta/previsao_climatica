```python
def run_notebook(notebook_path: list):
    """
    Executa um notebook Databricks a partir do caminho especificado.

    Args:
        notebook_path (list): Lista contendo o caminho do notebook a ser executado.

    Returns:
        str: O resultado da execução do notebook, se bem-sucedida.
        None: Caso ocorra um erro na execução, um erro é impresso e None é retornado.
    
    Raises:
        Exception: Captura e imprime qualquer erro ocorrido durante a execução do notebook.
    """
    try:
        return dbutils.notebook.run(notebook_path, 600)
    except Exception as e:
        print(f"Erro ao executar {notebook_path}: {e}")
        return None
```
