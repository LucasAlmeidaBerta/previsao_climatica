### Definições Gerais: 
Este notebook define os caminhos de armazenamento e organização do projeto de ingestão e tratamento de dados meteorológicos no Azure Data Lake, utilizando uma arquitetura em camadas (Bronze, Silver, Gold) para facilitar a governança, rastreabilidade e transformação dos dados.

---

### Informações Técnicas

### Objetivo: 
Definir, centralizar e padronizar os caminhos de armazenamento e schemas para dados meteorológicos em um ambiente Lakehouse utilizando Azure Data Lake.
 - Ingestão de dados de diferentes fontes de previsão do tempo (INPE, Weather API, Climatempo, Weather e Weather Map)
 - Definir caminhos para testes e validações com dados fictícios.
 - Gerenciar caminhos de schemas e tabelas no metastore.

 ### Armazenamento em camadas conforme a qualidade dos dados:
 - **LandingZone**: Dados brutos recém-coletados.
 - **Bronze**: Dados persistidos sem tratamento.
 - **Silver**: Dados limpos e normalizados.
 - **Gold**: Dados prontos para análise e consumo por dashboards/modelos.
