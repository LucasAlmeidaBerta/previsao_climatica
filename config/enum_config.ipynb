
# Projeto: Cadastro de Variáveis Globais

### Descrição Geral
Este documento define e centraliza as variáveis globais utilizadas no pipeline de ingestão e enriquecimento de dados meteorológicos. São incluídas as chaves de acesso a APIs externas, URLs utilizadas para scraping, e informações geográficas das capitais brasileiras necessárias para o processamento dos dados.

---

### Variáveis Cadastradas

| Tipo                    | Nome da Variável           | Descrição                                                                 |
|-------------------------|----------------------------|---------------------------------------------------------------------------|
| API Key                 | `WEATHER_API_KEY`          | Chave de acesso à API de previsão do tempo de uma fonte externa.         |
| API Key                 | `OPEN_WEATHER_MAP_KEY`     | Chave de acesso à API da OpenWeatherMap.                                 |
| URL de Webscraping      | `URL_WEATHER`              | URL base do site Weather.com para coleta de dados via scraping.          |
| Dados Geográficos       | `CAPITAIS_BRASIL`          | Dicionário com UF, nome da capital e suas coordenadas (lat, long).       |
| IDs para Scraping       | `IDS_CIDADES_WEATHER`      | Dicionário com UF, nome da capital e ID necessário para scraping.        |
| Cidades Acentuadas      | `NOME_CAPITAIS_CORRIGIDO`  | Dicionário com nomes das cidades sem acento para correção|
| API Key                 | `CHAVE_API_ROCKET`          | Token para o Rocket.         |
| API Key                 | `SENHA_APP_GMAIL`          | Token para o GMAIL.         |
| Endereço de email                | `EMAIL_REMETENTE`          | Endereço do GMAIL para envios automatizados de boletins climáticos.         |
