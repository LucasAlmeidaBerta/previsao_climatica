```python
"""
Executa a validação da tabela 'inpe' na camada bronze, utilizando Great Expectations,
e separa os dados validados e não validados (quarentena), salvando-os em formato Delta 
em seus respectivos diretórios.

Processo:
1. Chama a função `validar_tabelas_bronze` com os parâmetros do catálogo, camada, tabela
   e caminho para o contexto do Great Expectations.
2. Recupera os DataFrames resultantes da validação:
   - `df_validado`: registros que passaram em todas as regras.
   - `df_quarentena`: registros que falharam em alguma validação.
3. Salva os dados validados e de quarentena separadamente em formato Delta Lake.

Parâmetros:
- catalogo (str): Nome do catálogo do Unity Catalog.
- camada (str): Nome da camada (ex: 'bronze').
- tabelas (list): Lista com o nome das tabelas a serem validadas.
- contexto_path (str): Caminho para o contexto do Great Expectations.

Nota:
- Essa abordagem evita a conversão para Pandas, preservando a computação distribuída com Spark.
- Warnings gerados durante o processo são relacionados a bibliotecas de terceiros e não afetam o resultado final da execução.
"""
pasta_logs = lakehouse_logs_gx + "/bronze/inpe"

resultado_validacao = validar_tabelas_bronze(
    catalogo="grupo01",
    camada="bronze",
    tabelas=["inpe"],
    contexto_path=caminho_ge,
    pasta_logs=pasta_logs
)

df_inpe_validado = resultado_validacao["grupo01.bronze.inpe"]["df_validado"]
df_inpe_quarentena = resultado_validacao["grupo01.bronze.inpe"]["df_quarentena"]

pasta_quarentena = lakehouse_logs_gx + "/bronze/inpe/quarentena/inpe"
pasta_validos = lakehouse_logs_gx + "/bronze/inpe/validos/inpe"

df_inpe_validado = df_inpe_validado.dropDuplicates(["data_carga", "cidade", "estado", "temperatura_maxima", "temperatura_minima", "precipitacao", "fonte"])
df_inpe_quarentena = df_inpe_quarentena.dropDuplicates( ["data_carga", "cidade", "estado", "temperatura_maxima", "temperatura_minima", "precipitacao", "fonte"])

escreve_com_merge_delta(
    dataframe=df_inpe_validado,
    caminho_tabela_delta=pasta_validos,
    campos_chave=["data_carga", "cidade", "estado", "temperatura_maxima", "temperatura_minima", "precipitacao", "fonte"]
)
escreve_com_merge_delta(
    dataframe=df_inpe_quarentena,
    caminho_tabela_delta=pasta_quarentena,
    campos_chave=["data_carga", "cidade", "estado", "temperatura_maxima", "temperatura_minima", "precipitacao", "fonte"]
)
```
